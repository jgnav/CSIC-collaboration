```python
import numpy as np
import matplotlib.pyplot as plt
import tifffile
import os
```

```python
image_path = 'dataset/SEGMENTACION/annotations0-data0.tif'

if os.path.exists(image_path):
    # Load the image using tifffile.imread
    image = tifffile.imread(image_path)
    print(f"Image loaded successfully.")
    print(f"Shape: {image.shape}")
    print(f"Data Type: {image.dtype}")
    
    # Check if image is 3D
    if image.ndim == 3:
        print("Image is a 3D volume. Will use the middle slice for visualization.")
        # We will visualize the middle slice
        middle_slice_idx = image.shape[0] // 2
        image_to_show = image[middle_slice_idx]
    else:
        # Assuming 2D image
        image_to_show = image
else:
    print(f"File not found: {image_path}")
    # Create a dummy image for demonstration purposes if file doesn't exist
    print("Creating a dummy image for demonstration...")
    image = np.random.rand(256, 256).astype(np.float32)
    # Add some structure
    image[50:150, 50:150] = 0.8
    image[100:200, 100:200] = 0.2
    image_to_show = image
```

```python
# Binarize the image
# You can adjust the threshold value based on your image intensity range
threshold_value = 0.5  # Example threshold for normalized image (0-1)

# Create binary mask
binary_mask = (image > threshold_value).astype(np.uint8)

if binary_mask.ndim == 3:
    middle_slice_idx = binary_mask.shape[0] // 2
    binary_mask_to_show = binary_mask[middle_slice_idx]
else:
    binary_mask_to_show = binary_mask

print(f"Binary mask created with threshold: {threshold_value}")
print(f"Unique values in mask: {np.unique(binary_mask)}")
```

```python
# Visualize the results
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Check if image_to_show is defined (from previous cell), if not use image
if 'image_to_show' not in locals():
    # Fallback to middle slice if image is 3D and image_to_show wasn't set
    if image.ndim == 3:
        image_to_show = image[image.shape[0] // 2]
    else:
        image_to_show = image

if 'binary_mask_to_show' not in locals():
    # Fallback for binary mask
    if binary_mask.ndim == 3:
         binary_mask_to_show = binary_mask[binary_mask.shape[0] // 2]
    else:
         binary_mask_to_show = binary_mask

# Original Image
axes[0].imshow(image_to_show, cmap='gray')
axes[0].set_title('Original Image (Slice)')
axes[0].axis('off')

# Binary Mask
axes[1].imshow(binary_mask_to_show, cmap='gray')
axes[1].set_title('Binarized Mask (Slice)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
